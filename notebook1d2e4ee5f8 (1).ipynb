```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

MODEL_NAME = 'microsoft/Phi-3-mini-4k-instruct'

print("Loading Phi-3-mini (~2-4 mins)...")
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    device_map='auto',
    trust_remote_code=True,
    torch_dtype=torch.float16,
    attn_implementation='eager'
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = 'right'

print(f"✅ Model loaded!")
print(f"   Parameters : {model.num_parameters():,}")
print(f"   GPU memory : {torch.cuda.memory_allocated()/1e9:.2f} GB")
```

```python
import json
from datasets import Dataset

train_data, eval_data = [], []
with open('/kaggle/working/bioinformagus/Fine-tune/input_data/train_1.jsonl') as f:
    for line in f: train_data.append(json.loads(line))
with open('/kaggle/working/bioinformagus/Fine-tune/input_data/test_1.jsonl') as f:
    for line in f: eval_data.append(json.loads(line))

def format_sample(sample):
    messages = sample.get('messages', [])
    text = ''
    for msg in messages:
        role    = msg.get('role', '')
        content = msg.get('content', '')
        if role == 'user':
            text += f'<|user|>\n{content}<|end|>\n'
        elif role == 'assistant':
            text += f'<|assistant|>\n{content}<|end|>\n'
    return {'text': text}

train_dataset = Dataset.from_list([format_sample(s) for s in train_data])
eval_dataset  = Dataset.from_list([format_sample(s) for s in eval_data])
print(f"✅ Train: {len(train_dataset)} | Eval: {len(eval_dataset)}")
```

```python
from peft import LoraConfig, get_peft_model, TaskType

model.enable_input_require_grads()

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=['q_proj', 'k_proj', 'v_proj', 'o_proj'],
    lora_dropout=0.05,
    bias='none',
    task_type=TaskType.CAUSAL_LM
)

model = get_peft_model(model, lora_config)

for name, param in model.named_parameters():
    if param.requires_grad:
        param.data = param.data.float()

model.print_trainable_parameters()
print("✅ LoRA ready!")
```

```python
from trl import SFTTrainer
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir='/kaggle/working/phi3_bioagent_finetuned',
    num_train_epochs=3,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    learning_rate=2e-4,
    fp16=False,
    bf16=False,
    logging_steps=10,
    eval_strategy='epoch',
    save_strategy='epoch',
    load_best_model_at_end=True,
    warmup_ratio=0.03,
    lr_scheduler_type='cosine',
    report_to='none',
    optim='adamw_torch',
    gradient_checkpointing=True,
    gradient_checkpointing_kwargs={'use_reentrant': False},
)

trainer = SFTTrainer(
    model=model,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    args=training_args,
    dataset_text_field='text',
    max_seq_length=512,
)

print("✅ Starting fine-tuning...")
print("Expected: ~1 hour on T4 x2\n")
trainer.train()
print("\n✅ Fine-tuning complete!")
```

```python
ADAPTER_PATH = '/kaggle/working/phi3_bioagent_adapter'
trainer.model.save_pretrained(ADAPTER_PATH)
tokenizer.save_pretrained(ADAPTER_PATH)
print(f"✅ Saved! Files: {os.listdir(ADAPTER_PATH)}")
```

```python
import json, faiss, pickle, torch, numpy as np
from transformers import AutoTokenizer as AT, AutoModel as AM

# Load nfc.jsonl documents
with open('/kaggle/working/bioinformagus/RAG/nfc.jsonl', 'rb') as f:
    raw = f.read().decode('utf-8', errors='ignore')

raw = raw.strip().lstrip('[').rstrip(']')
parts = raw.split('}\n{')

docs = []
for i, part in enumerate(parts):
    try:
        if i == 0:                obj = json.loads(part + '}')
        elif i == len(parts)-1:   obj = json.loads('{' + part)
        else:                     obj = json.loads('{' + part + '}')
        docs.append(obj)
    except: pass

print(f"✅ Documents loaded: {len(docs)}")

# Build embeddings
EMBED_MODEL = 'sentence-transformers/all-MiniLM-L6-v2'
embed_tok = AT.from_pretrained(EMBED_MODEL)
embed_mod = AM.from_pretrained(EMBED_MODEL).to('cuda')
embed_mod.eval()

def get_embedding(texts):
    inputs = embed_tok(texts, padding=True, truncation=True,
                       max_length=256, return_tensors='pt').to('cuda')
    with torch.no_grad():
        out = embed_mod(**inputs)
    return out.last_hidden_state.mean(dim=1).cpu().float().numpy()

# Build chunks
chunks = []
for d in docs:
    content = f"Question: {d['question']}\nAnswer: {d['answer']}"
    chunks.append(content[:512])

print(f"Building embeddings for {len(chunks)} documents...")
all_emb = []
for i in range(0, len(chunks), 32):
    all_emb.extend(get_embedding(chunks[i:i+32]))
    if i % 320 == 0:
        print(f"  Progress: {i}/{len(chunks)}")

matrix = np.array(all_emb).astype('float32')
index  = faiss.IndexFlatL2(matrix.shape[1])
index.add(matrix)
print(f"✅ FAISS index built: {index.ntotal} vectors")

# Save
faiss.write_index(index, '/kaggle/working/faiss_index.bin')
with open('/kaggle/working/chunks.pkl', 'wb') as f:
    pickle.dump(chunks, f)
print("✅ RAG saved!")
```

```python
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer

MODEL_NAME   = 'microsoft/Phi-3-mini-4k-instruct'
ADAPTER_PATH = '/kaggle/working/phi3_bioagent_adapter'

print("Loading fine-tuned model...")
base_model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME, device_map='auto',
    trust_remote_code=True, torch_dtype=torch.float16,
    attn_implementation='eager'
)
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
agent1_model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)
agent1_model.eval()
print(f"✅ Agent 1 loaded! GPU: {torch.cuda.memory_allocated()/1e9:.2f} GB")

def run_agent1(question):
    prompt = f'<|user|>\n{question}<|end|>\n<|assistant|>\n'
    inputs = tokenizer(prompt, return_tensors='pt').to('cuda')
    L = inputs['input_ids'].shape[1]
    with torch.no_grad():
        out = agent1_model.generate(**inputs, max_new_tokens=256,
                                    temperature=0.7, do_sample=True,
                                    pad_token_id=tokenizer.eos_token_id)
    return tokenizer.decode(out[0][L:], skip_special_tokens=True).strip()

def run_agent2(question):
    q_emb = get_embedding([question]).astype('float32')
    _, idx = index.search(q_emb, 3)
    return '\n\n'.join([chunks[i] for i in idx[0]])[:800]

def run_reasoning_agent(question, a1, a2):
    prompt = (f'<|user|>\nYou are a bioinformatics expert.\n\n'
              f'Question: {question}\n\n'
              f'Knowledge 1: {a1[:300]}\n\n'
              f'Knowledge 2: {a2[:300]}\n\n'
              f'Give a clear 2-3 sentence expert answer.<|end|>\n<|assistant|>\n')
    inputs = tokenizer(prompt, return_tensors='pt').to('cuda')
    L = inputs['input_ids'].shape[1]
    with torch.no_grad():
        out = agent1_model.generate(**inputs, max_new_tokens=200,
                                    temperature=0.3, do_sample=True,
                                    pad_token_id=tokenizer.eos_token_id)
    return tokenizer.decode(out[0][L:], skip_special_tokens=True).strip()

def bioagents_ask(question):
    print(f"\n{'='*60}")
    print(f"QUESTION: {question}")
    print(f"{'='*60}")
    print("\n🤖 Agent 1 (Fine-tuned)...")
    a1 = run_agent1(question)
    print(f"Agent 1: {a1}\n")
    print("🔍 Agent 2 (RAG)...")
    a2 = run_agent2(question)
    print(f"Agent 2: {a2[:200]}\n")
    print("🧠 Reasoning Agent...")
    final = run_reasoning_agent(question, a1, a2)
    print(f"\n{'='*60}")
    print("✅ FINAL ANSWER:")
    print(f"{'='*60}")
    print(final)
    return final

print("✅ All 3 agents ready!")
```

```python
import pandas as pd
from rouge_score import rouge_scorer

df = pd.read_csv('/kaggle/working/bioinformagus/Evaluation/analysis_and_tools_only.csv')
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

results = []
for i, row in df.head(5).iterrows():
    question      = str(row.get('title', ''))
    expert_answer = str(row.get('answer_content', ''))
    a1     = run_agent1(question)
    a2     = run_agent2(question)
    answer = run_reasoning_agent(question, a1, a2)
    scores = scorer.score(expert_answer, answer)
    results.append({
        'question': question[:60],
        'rouge1'  : round(scores['rouge1'].fmeasure, 3),
        'rouge2'  : round(scores['rouge2'].fmeasure, 3),
        'rougeL'  : round(scores['rougeL'].fmeasure, 3),
    })
    print(f"Q{i+1}: ROUGE-1={scores['rouge1'].fmeasure:.3f}")

results_df = pd.DataFrame(results)
print("\n" + "="*50)
print("EVALUATION SUMMARY")
print("="*50)
print(results_df.to_string())
print(f"\nAverage ROUGE-1 : {results_df['rouge1'].mean():.3f}")
print(f"Average ROUGE-2 : {results_df['rouge2'].mean():.3f}")
print(f"Average ROUGE-L : {results_df['rougeL'].mean():.3f}")
```
